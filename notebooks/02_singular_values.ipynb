### Singular Value Decay Across Test Images

```python
import sys
from pathlib import Path
import matplotlib.pyplot as plt

from src.compression import truncated_components
from src.io_utils import load_grayscale
```

```python
PROJECT_ROOT = Path.cwd().parent
sys.path.append(str(PROJECT_ROOT))
```

```python
image_path = PROJECT_ROOT / "data" / "raw" / "human.jpg"
image = load_grayscale(str(image_path))
```

```python
_, S, _ = truncated_components(image, min(image.shape))
```

```python
plt.figure(figsize=(8, 5))
plt.semilogy(S)
plt.xlabel('Index j')
plt.ylabel('Singular value σⱼ (log scale)')
plt.title('Singular value decay')
plt.show()
```

```python
image_files = sorted((PROJECT_ROOT / "data" / "raw").glob("*.jpg"))

plt.figure(figsize=(8, 5))
for image_path in image_files:
    img = load_grayscale(str(image_path))
    _, S, _ = truncated_components(img, min(img.shape))
    plt.semilogy(S, label=image_path.name)

plt.xlabel('Index j')
plt.ylabel('Singular value σⱼ (log scale)')
plt.title('Singular value decay across test images')
plt.legend()
plt.show()
```
